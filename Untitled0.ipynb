```python
!pip install yfinance FinMind
import yfinance as yf
from FinMind.data import DataLoader
from datetime import datetime, timedelta

def get_tsmc_data():
    print("--- 晨間台積電 (TSMC) 數據彙整 ---")
    print(f"執行日期: {datetime.now().strftime('%Y-%m-%d')}\n")

    # 1. 抓取股價 (使用 yfinance)
    try:
        tsmc_tw = yf.Ticker("2330.TW")
        close_tw = tsmc_tw.history(period="1d")['Close'].iloc[0]
        print(f"📍 台積電 (2330.TW) 最新收盤價: {close_tw:.2f} 元")

        tsmc_adr = yf.Ticker("TSM")
        close_adr = tsmc_adr.history(period="1d")['Close'].iloc[0]
        print(f"📍 台積電 ADR (TSM) 最新收盤價: {close_adr:.2f} 美元")
    except Exception as e:
        print("股價讀取失敗")

    # 2. 抓取籌碼 (使用 FinMind 串接台灣證交所)
    print("\n📍 法人與主力籌碼動態:")
    try:
        dl = DataLoader()
        # 抓取過去 7 天的資料，確保能避開六日，抓到最近一個交易日的數據
        start_date = (datetime.now() - timedelta(days=7)).strftime('%Y-%m-%d')

        # 外資買賣超
        df_inst = dl.taiwan_stock_institutional_investors(stock_id='2330', start_date=start_date)
        if not df_inst.empty:
            # 篩選外資並抓最後一天，將股數除以 1000 換算成「張」
            foreign_data = df_inst[df_inst['name'] == 'Foreign_Investor'].tail(1)
            net_buy_sell = (foreign_data['buy'].values[0] - foreign_data['sell'].values[0]) / 1000
            date_inst = foreign_data['date'].values[0]
            print(f"- ({date_inst}) 外資買賣超: {net_buy_sell:,.0f} 張")
        else:
            print("- 外資買賣超: 目前無最新資料")

        # 散戶融資餘額
        df_margin = dl.taiwan_stock_margin_purchase_short_sale(stock_id='2330', start_date=start_date)
        if not df_margin.empty:
            latest_margin = df_margin.tail(1)
            margin_bal = latest_margin['MarginPurchaseBalance'].values[0] / 1000
            date_margin = latest_margin['date'].values[0]
            print(f"- ({date_margin}) 散戶融資餘額: {margin_bal:,.0f} 張")
        else:
             print("- 散戶融資餘額: 目前無最新資料")

    except Exception as e:
        print(f"- 籌碼資料抓取失敗 (可能遇到證交所連線限制)")

    print("\n💡 提醒：籌碼與股價都到位了，綜合評估後再決定今日是否進場喔！")

if __name__ == "__main__":
    get_tsmc_data()
```
